```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('amazon.csv')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
# 1. Clean prices: remove the ₹ symbol and commas, then convert to float

df['discounted_price'] = df['discounted_price'].str.replace('₹', '').str.replace(',', '').astype(float)
df['actual_price'] = df['actual_price'].str.replace('₹', '').str.replace(',', '').astype(float)
```

```python
# 2. Clean discount percentage: remove the % sign and convert to float (e.g. 50% becomes 0.50)

df['discount_percentage'] =  df['discount_percentage'].str.replace('%', '').astype(float) / 100
```

```python
# 3. Clean rating count: remove commas and convert to numeric
df['rating_count'] = df['rating_count'].str.replace(',', '')
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')
```

```python
df['rating_count'].value_counts().sum()
```

```python
df.columns
```

```python
# Drop the handful of missing rows in ratings and rating_count
df = df.dropna(subset=['rating', 'rating_count']).copy()
```

```python
# Split the string at '|' and grab the first chunk (Main Category)
df['main_category'] = df['category'].str.split('|').str[0]

# Grab the second chunk (Sub Category)
df['sub_category'] = df['category'].str.split('|').str[1]
```

```python
# Add spaces around the '&' so "Computers&Accessories" becomes "Computers & Accessories"
df['main_category'] = df['main_category'].str.replace('&', ' & ')
df['sub_category'] = df['sub_category'].str.replace('&', ' & ')
```

```python
# Drop the original messy column category
df = df.drop(columns=['category'])
```

```python
print("--- Top 5 Main Categories ---")
print(df['main_category'].value_counts().head(5))
```

```python
# Calculate the exact amount saved in ₹
df['discount_amount_inr'] = df['actual_price'] - df['discounted_price']
```

```python
# Define the price buckets (0-1000, 1000-5000, 5000+)\
bins = [0, 1000, 5000, float('inf')]
labels = ['Budge (Under 1k)', 'Mid-Range (1k - 5k)', 'Premium (5K+)']
```

```python
# Apply the buckets using pd.cut()
df['price_tier'] = pd.cut(df['discounted_price'], bins=bins, labels=labels)
```

```python
print("--- Products by Price Tier ---")
print(df['price_tier'].value_counts())
```

```python
df.columns
```

```python
columns_to_drop = [
    'about_product', 'review_content', 'img_link', 'product_link', 
    'user_name', 'product_name', 'review_title', 'user_id', 'review_id'
]
df_clean = df.drop(columns=columns_to_drop).copy()
```

```python
final_filename = 'amazon_india_dashboard_ready.csv'
df_clean.to_csv(final_filename, index=False)
print(f"Export Complete: {final_filename}")
print(f"Final Rows: {len(df_clean)}")
print(f"Final Columns: {len(df_clean.columns)}")
```
